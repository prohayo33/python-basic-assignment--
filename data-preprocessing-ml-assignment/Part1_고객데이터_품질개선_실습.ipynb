```python
# 2. 날짜형 자료형 변환
 # 변환하기 전에 형태 확인하기 -> . / - 일/월/년 형식 등 (표기 형식 다양해서 통일화 먼저 필요)
for col in ['LastPurchaseDate', 'SignupDate']:
    print(f'\n===== {col} =====')

    s = df_clean[col].dropna().astype(str)

    patterns = {
        'YYYY-MM-DD': r'^\d{4}-\d{1,2}-\d{1,2}$',
        'YYYY.MM.DD': r'^\d{4}\.\d{1,2}\.\d{1,2}$',
        'YYYY/MM/DD': r'^\d{4}/\d{1,2}/\d{1,2}$',
        'DD-MM-YYYY': r'^\d{1,2}-\d{1,2}-\d{4}$',
        'DD.MM.YYYY': r'^\d{1,2}\.\d{1,2}\.\d{4}$',
        'DD/MM/YYYY': r'^\d{1,2}/\d{1,2}/\d{4}$',
    }

    for name, pattern in patterns.items():
        count = s.str.match(pattern).sum()
        print(f'{name}: {count}개')

# 날짜 컬럼
date_cols = ['LastPurchaseDate', 'SignupDate']

for col in date_cols:
    # 날짜 구분자를 '-'로 통일
    df_clean[col] = (
        df_clean[col]
        .astype('string')
        .str.replace('.', '-', regex=False)
        .str.replace('/', '-', regex=False)
    )

# LastPurchaseDate
df_clean['LastPurchaseDate'] = (
    df_clean['LastPurchaseDate']
    .astype('string')
    .str.replace('.', '-', regex=False)
    .str.replace('/', '-', regex=False)
)

df_clean['LastPurchaseDate'] = pd.to_datetime(
    df_clean['LastPurchaseDate'],
    format='%Y-%m-%d',
    errors='coerce'
)


# SignupDate
df_clean['SignupDate'] = (
    df_clean['SignupDate']
    .astype('string')
    .str.replace('.', '-', regex=False)
    .str.replace('/', '-', regex=False)
)

# DD-MM-YYYY → YYYY-MM-DD
mask = df_clean['SignupDate'].str.match(
    r'^\d{1,2}-\d{1,2}-\d{4}$',
    na=False
)

df_clean.loc[mask, 'SignupDate'] = (
    pd.to_datetime(
        df_clean.loc[mask, 'SignupDate'],
        format='%d-%m-%Y',
        errors='coerce'
    ).dt.strftime('%Y-%m-%d')
)

# 최종 날짜형 변환
df_clean['SignupDate'] = pd.to_datetime(
    df_clean['SignupDate'],
    format='%Y-%m-%d',
    errors='coerce'
)

# 자료형 잘 변환되었는지 확인하기
print(f'수치형 자료형 변환: {df_clean[['Age','AveragePurchaseAmount']].dtypes}')
print(f'날짜형 자료형 변환: {df_clean[['LastPurchaseDate', 'SignupDate']].dtypes} ')

# 3. 범주 표기 정리하기
# Gender, Region, MembershipLevel, CouponUsed, PreferredCategory
# Gender 
df_clean['Gender'] = df_clean['Gender'].str.strip().replace({
    'female':'Female',
    'FEMALE': 'Female',
    'F': 'Female',
    'male': 'Male',
    'MALE': 'Male',
    'M': 'Male',

})

# Region - unique 찍어본 결과 대소문자, 띄어쓰기 문제
df_clean['Region'] = df_clean['Region'].str.strip().str.title()

#MembershipLevel
df_clean['MembershipLevel'] = (
    df_clean['MembershipLevel']
    .str.strip()
    .str.title()
    .replace({'Vip': 'VIP'})
)

# CouponUsed
df_clean['CouponUsed'] = df_clean['CouponUsed'].str.strip().replace({
    'No': 'N',
    'n': 'N',
    'Yes': 'Y',
    'y': 'Y'
})

#PreferredCategory
df_clean['PreferredCategory'] = (
    df_clean['PreferredCategory']
    .str.strip()
    .str.title()
)

# 범주형 데이터 제대로 통일되었는지 확인해 보기
for col in ['Gender', 'Region', 'MembershipLevel',
            'CouponUsed', 'PreferredCategory']:
    print(f'\n[{col}]')
    display(df_clean[col].value_counts(dropna=False))
```

```python
# TODO: 결측치와 전체 행 기준 중복 데이터를 처리하세요.
 # 1. 전체 칼럼의 값이 완전히 동일한 중복 행인지 확인하기
duplicated_df = df_clean[df_clean.duplicated(keep=False)]
duplicated_df = duplicated_df.sort_values(by='CustomerID', ascending=False)
display(duplicated_df)

# 1.2 실제로  완전히 중복된 값이란거 확인 -> 중복 데이터 첫번째 등장한 행만 유지
df_clean = df_clean.drop_duplicates(keep='first')

# 1.3 잘 삭제되었는지 확인
print('중복 데이터 삭제 후: ',df_clean.duplicated().sum())

df_clean['Age'] = df_clean['Age'].astype(float)
df_clean['AveragePurchaseAmount'] = df_clean['AveragePurchaseAmount'].astype(float)


# 2. 결측치 처리하기
# 결측치가 있는 행:   Gender, Age, Region, MembershipLevel, AveragePurchaseAmount, SatisfactionScore, LastPurchaseDate, PreferredCategory 
# 2.1 이상치가 있을 가능성이 있는 수치형 변수 - 중앙값으로 대체하기
#  - 이상치 확인이 필요한 컬럼: Age, VisitCount,AveragePurchaseAmount, TotalPurchaseAmount
#  - membershiplevel, PreferredCategory로 유의미한 차이가 있는지 확인하기
print('membership,category 선호도별로 결측치 대체할 만한 유의미한 차이가 있는지 확인')
display(df_clean.groupby(
    ['MembershipLevel', 'PreferredCategory']
)[
    ['VisitCount','AveragePurchaseAmount', 'TotalPurchaseAmount', 'Age']
].agg(['count', 'mean', 'median']))

# 2.1.1 MembershipLevel + PreferredCategory별 중앙값으로 결측치 대체
group_cols = ['MembershipLevel', 'PreferredCategory']

for col in [
    'VisitCount',
    'AveragePurchaseAmount',
    'TotalPurchaseAmount',
    'Age'
]:
    df_clean[col] = df_clean[col].fillna(
        df_clean.groupby(group_cols)[col].transform('median')
    )

# 2.1.2. 위 그룹 기준으로도 채워지지 않은 결측치는 전체 중앙값으로 대체
for col in [
    'VisitCount',
    'AveragePurchaseAmount',
    'TotalPurchaseAmount',
    'Age'
]:
    df_clean[col] = df_clean[col].fillna(
        df_clean[col].median()
    )


# 2.2 이상치 확률이 적은 수치형 변수 - 평균으로 대체하기
# - satisfactionscore
df_clean['SatisfactionScore'] = df_clean['SatisfactionScore'].fillna(
    df_clean['SatisfactionScore'].mean()
)

# 2.3 범주형 변수: 'Gender', 'Region', 'MembershipLevel','CouponUsed', 'PreferredCategory'
 # 최빈값으로 대체하기
 # or NaN 값을 Unknown으로 처리하기 (결측 자체가 하나의 정보일 수 있다는 의미) 
 # -> 'Gender'나 'Region', 'CouponUsed','PreferredCategory' 최빈값을 대체하는 거 의미 없음 -> Unknow 값으로 처리하기
categorical_cols = [
    'Gender',
    'Region',
    'CouponUsed',
    'PreferredCategory'
]

for col in categorical_cols:
    df_clean[col] = df_clean[col].fillna('Unknown')

# membershiplevel은 TotalPurchaseAmount랑 비교해서 유의미한 값인지 확인 후 -> 대체
display(df_clean.groupby('MembershipLevel')[
    ['TotalPurchaseAmount']
].agg(['count', 'mean', 'median'])
)

# 1. MembershipLevel별 TotalPurchaseAmount의 중앙값 계산
membership_median = df_clean.groupby('MembershipLevel')[
    'TotalPurchaseAmount'
].median()

#2. 등급 순서대로 정렬
membership_median = membership_median.reindex(
    ['Basic', 'Silver', 'Gold', 'VIP']
)

print(membership_median)

#3. 등급 사이의 구간 경계값 자동 계산
thresholds = [
    (membership_median.iloc[i] + membership_median.iloc[i + 1]) / 2
    for i in range(len(membership_median) - 1)
]

print(thresholds)

#4. 계산된 구간 이용해서 결측치 대체하기
def fill_membership_level(row):
    if pd.notna(row['MembershipLevel']):
        return row['MembershipLevel']

    amount = row['TotalPurchaseAmount']

    if amount < thresholds[0]:
        return 'Basic'
    elif amount < thresholds[1]:
        return 'Silver'
    elif amount < thresholds[2]:
        return 'Gold'
    else:
        return 'VIP'


df_clean['MembershipLevel'] = df_clean.apply(
    fill_membership_level,
    axis=1
)

# 날짜형 결측치 삭제
df_clean = df_clean.dropna(subset=['LastPurchaseDate'])



# TODO: 처리 전후 결과를 확인하세요.
# 컬럼별 결측치 개수 확인
print('\n전체 결측치 개수:', df_clean.isna().sum().sum())
```

```python
# TODO: Q1, Q3, IQR, 하한과 상한을 반환하세요.
def detect_outliers_iqr(df, col, id_col="CustomerID", show=5):
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = (df[col] < lower) | (df[col] > upper)
    count = outliers.sum()

    print(f"--- {col} ---")
    print(f"Q1: {q1}, Q3: {q3}, IQR: {iqr}")
    print(f"하한: {lower}, 상한: {upper}")
    print(f"이상치 후보 수: {count}")

    result = df.loc[outliers, [id_col, col]]
    display(result.head(show))  # 노트북이면 display, 아니면 print(result.head(show))

    return {
        "col": col,
        "q1": q1,
        "q3": q3,
        "iqr": iqr,
        "lower": lower,
        "upper": upper,
        "outlier_mask": outliers,
        "outlier_count": count
    }
```

```python
cols_to_check = [
    "Age",
    "VisitCount",
    "AveragePurchaseAmount",
    "TotalPurchaseAmount",
    "SatisfactionScore"
]

results = {}

for col in cols_to_check:
    results[col] = detect_outliers_iqr(df_clean, col)
```

```python
# 경계값 조정하기
def clip_outliers(df_before, df_after, col, lower, upper, show=5):
    # float64로 변환 후 클리핑
    df_after[col] = df_after[col].astype("float64")
    df_after[col] = df_after[col].clip(lower=lower, upper=upper)

    # 변경된 값 비교
    changed = df_before[col] != df_after[col]
    print(f"--- {col} ---")
    print("변경한 값 수:", changed.sum())

    comparison = pd.DataFrame()
    comparison["처리 전"] = df_before.loc[changed, col]
    comparison["처리 후"] = df_after.loc[changed, col]
    display(comparison.head(show))

    print(f"{col} value_counts:")
    display(df_after[col].value_counts(dropna=False))
    print()

    return changed
```

```python
df_before_outlier = df_clean.copy()

cols_to_check = ["Age", "VisitCount", "AveragePurchaseAmount", "TotalPurchaseAmount", "SatisfactionScore"]

changed_masks = {}
for col in cols_to_check:
    lower = results[col]["lower"]
    upper = results[col]["upper"]
    changed_masks[col] = clip_outliers(df_before_outlier, df_clean, col, lower, upper)
```
