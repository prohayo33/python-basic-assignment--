## 사전 준비 2. 자동차 판매 CSV 파일 불러오기

### 배경

환경 구축을 마친 뒤 영업관리팀에서 전달받은 자동차 판매 CSV 데이터를 확인합니다.

분석에 앞서 CSV 파일을 Python에서 다룰 수 있는 형태로 불러오고,
이후 조건문·반복문을 활용한 판매 분석에 사용할 데이터를 준비합니다.

### 요구사항

1. `csv` 모듈을 불러오세요.
2. `자동차_판매_데이터.csv` 파일을 UTF-8 형식으로 여세요.
3. `csv.DictReader()`를 사용하여 각 행을 딕셔너리 형태로 읽으세요.
4. 전체 데이터를 리스트로 변환하여 `sales_data` 변수에 저장하세요.
5. 전체 거래 수와 첫 번째 판매 기록을 출력하세요.

### ✅ 확인 포인트

- CSV 파일이 정상적으로 불러와져야 합니다.
- 전체 판매 데이터는 리스트 형태로 저장되어야 합니다.
- 리스트 내부의 각 판매 기록은 딕셔너리 형태여야 합니다.
- 이후 Part 1~3에서 `sales_data`를 사용할 수 있어야 합니다.


```python
# 여기에 코드를 작성하세요.
import csv

with open("자동차_판매_데이터.csv", encoding='utf-8') as file:
    reader = csv.DictReader(file)
    sales_data = list(reader)


print("전체 거래 수:", len(sales_data))
print("sales data의 자료형:", type(sales_data))
print("sales_data의 첫 번째 판매 기록의 자료형: ", type(sales_data[1]))
print("첫 번째 판매 기록: ", sales_data[0])
```
