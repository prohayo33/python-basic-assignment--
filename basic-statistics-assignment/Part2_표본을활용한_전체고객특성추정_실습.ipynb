## 제공 코드. 모집단 데이터 준비

아래 코드는 준비 코드이며 평가 대상이 아닙니다.

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_FILE = "online_shopping_customer_data.csv"

df = pd.read_csv(DATA_FILE)

population_columns = ["PurchaseAmount", "MonthlyVisitCount"]

population_df = df.dropna(subset=population_columns).copy()

print("전체 고객 수:", len(df))
print("분석에 사용할 모집단 고객 수:", len(population_df))

population_df[population_columns].head()
```

# 필수 문제 1. 모집단의 평균 구매 특성 확인

## 업무 상황

분석팀은 표본을 추출하기 전에 전체 고객의 평균 구매금액과 평균 방문횟수를 기준값으로 확인하려고 합니다.

## 요구사항

1. 모집단의 전체 고객 수를 출력합니다.
2. `PurchaseAmount`의 모집단 평균을 계산합니다.
3. `MonthlyVisitCount`의 모집단 평균을 계산합니다.
4. 계산 결과를 하나의 표로 정리합니다.
5. 모집단과 표본의 차이를 한 문장으로 작성합니다.


```python
# 필수 문제 1 코드를 작성하세요.

# 모집단 평균 계산
purchase_mean = population_df['PurchaseAmount'].mean().round(2)
visit_mean = population_df['MonthlyVisitCount'].mean().round(2)

# 결과를 하나의 표로 정리
population_summary = pd.DataFrame({
    '구분': ['전체 모집단 고객 수', '평균 PurchaseAmount', '평균 MonthlyVisitCount'],
    '값': [
        len(population_df),
        purchase_mean,
        visit_mean
    ]
})

population_summary
```

- 모집단과 표본의 차이: 모집단은 분석하고자 하는 전체 대상을 말하고 표본은 모집단에서 일부 데이터를 추출하는 것을 말한다.

# 필수 문제 2. 서로 다른 크기의 고객 표본 추출 및 비교

## 업무 상황

분석팀은 적은 수의 고객만 확인했을 때와 더 많은 고객을 확인했을 때
전체 고객의 평균 구매 특성을 얼마나 비슷하게 추정할 수 있는지 비교하려고 합니다.

## 요구사항

1. 모집단에서 고객 30명을 무작위로 추출합니다.
2. 모집단에서 고객 100명을 무작위로 추출합니다.
3. 두 표본은 코드를 다시 실행해도 같은 고객이 추출되도록 `random_state`를 설정합니다.
4. 각 표본의 평균 구매금액과 평균 방문횟수를 계산합니다.
5. 모집단 평균과 두 표본평균을 하나의 표로 정리합니다.
6. 모집단 평균과 각 표본평균의 차이를 계산합니다.

## 힌트

- `DataFrame.sample()`을 사용하여 표본을 추출할 수 있습니다.
- 평균 차이는 `표본평균 - 모집단 평균`으로 계산할 수 있습니다.
- 표본은 동일 고객이 중복되지 않도록 추출합니다.


```python
# 필수 문제 2 코드를 작성하세요.

# 1. 고객 30명 무작위 추출
sample_30 = population_df.sample(n=30, random_state=42)

# 2. 고객 100명 무작위 추출
sample_100 = population_df.sample(n=100, random_state=42)

# 3. 각 표본의 평균 계산
sample_30_purchase_mean = sample_30['PurchaseAmount'].mean().round(2)
sample_30_visit_mean = sample_30['MonthlyVisitCount'].mean().round(2)

sample_100_purchase_mean = sample_100['PurchaseAmount'].mean().round(2)
sample_100_visit_mean = sample_100['MonthlyVisitCount'].mean().round(2)

# 4. 모집단 평균
population_purchase_mean = population_df['PurchaseAmount'].mean().round(2)
population_visit_mean = population_df['MonthlyVisitCount'].mean().round(2)

# 5. 모집단 평균과 표본평균을 하나의 표로 정리
comparison = pd.DataFrame({
    '구분': ['모집단', '표본 30명', '표본 100명'],
    '평균 PurchaseAmount': [
        population_purchase_mean,
        sample_30_purchase_mean,
        sample_100_purchase_mean
    ],
    '평균 MonthlyVisitCount': [
        population_visit_mean,
        sample_30_visit_mean,
        sample_100_visit_mean
    ]
})

# 6. 모집단 평균과 각 표본평균의 차이
difference = pd.DataFrame({
    '구분': ['표본 30명', '표본 100명'],
    'PurchaseAmount 평균 차이': [
        sample_30_purchase_mean - population_purchase_mean,
        sample_100_purchase_mean - population_purchase_mean
    ],
    'MonthlyVisitCount 평균 차이': [
        sample_30_visit_mean - population_visit_mean,
        sample_100_visit_mean - population_visit_mean
    ]
})

display('[모집단 & 표본평균] ',comparison)
display('[모집단 평균과 표본평균 차이]',difference)
```
